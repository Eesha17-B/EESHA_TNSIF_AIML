1.Reverse an array

```python
arr = [10, 20, 30, 40, 50]

result = []

for i in range(len(arr) - 1, -1, -1):
    result.append(arr[i])

print(result)
```

2.Count even,odd and zero

```python
arr = [10, 5, 0, 7, 8, 0, 13, 4]

even = 0
odd = 0
zero = 0

for num in arr:
    if num == 0:
        zero += 1
    elif num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
print("Zero:", zero)
```

3.Sum of positive and negative numbers

```python
arr = [10, -5, 20, -8, 15, -2]

positive_sum = 0
negative_sum = 0

for num in arr:
    if num > 0:
        positive_sum += num
    elif num < 0:
        negative_sum += num

print("Positive Sum:", positive_sum)
print("Negative Sum:", negative_sum)
```

4.Remove duplicate elements

```python
arr = [10, 20, 10, 30, 20, 40, 30]

result = []

for num in arr:
    if num not in result:
        result.append(num)

print(result)
```

5.Find the missing number

```python
arr = [1, 2, 3, 5, 6, 7]

n = 7
total = n * (n + 1) // 2

for num in arr:
    total = total - num

print("Missing Number:", total)
```

6.Rotate an array

```python
arr = [1, 2, 3, 4, 5]
k = 2
k = k % len(arr)
result = arr[-k:] + arr[:-k]
print(result)
```

7.Most frequent number

```python
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

frequency = {}
for num in arr:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

most_frequent = max(frequency, key=frequency.get)

print("Most Frequent Element:", most_frequent)
print("Frequency:", frequency[most_frequent])
```

8.Maximim subarray sum

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = arr[0]
max_sum = arr[0]

start = 0
best_start = 0
best_end = 0

for i in range(1, len(arr)):
    if arr[i] > current_sum + arr[i]:
        current_sum = arr[i]
        start = i
    else:
        current_sum = current_sum + arr[i]

    if current_sum > max_sum:
        max_sum = current_sum
        best_start = start
        best_end = i

print("Maximum Subarray Sum:", max_sum)
print("Subarray:", arr[best_start:best_end + 1])
```

9.check whether two arrays are equal

```python
arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

if len(arr1) != len(arr2):
    print("Arrays are Not Equal")
else:
    count1 = {}
    count2 = {}

    for num in arr1:
        count1[num] = count1.get(num, 0) + 1

    for num in arr2:
        count2[num] = count2.get(num, 0) + 1

    if count1 == count2:
        print("Arrays are Equal")
    else:
        print("Arrays are Not Equal")
```
